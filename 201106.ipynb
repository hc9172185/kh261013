```python
import pandas as pd
titanic = pd.read_csv("data/raw/titanic.csv")

cond01 = titanic["pclass"] == 1
cond02 = titanic["fare"] >= 100
cond03 = titanic["sex"] == "female"
mask = cond01&cond02&cond03

result = titanic.loc[mask]

result_sorted = result.sort_values(by="fare", ascending=False)
result_sorted.head()

result["survived"].mean()
```
