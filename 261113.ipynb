```python
import pandas as pd
import glob
baemin_file = sorted(glob.glob("data/baemin/baemin*.csv"))
coupang_file = sorted(glob.glob("data/coupang/coupang*.csv"))

print("배민 파일:", baemin_file)
print("쿠팡 파일:", coupang_file)
```

```python
df_baemin = pd.read_csv(baemin_file[0])
df_baemin = df_baemin.rename(columns={
    "order_at": "주문일시",
    "total_bill": "전체금액",
    "ad_yn": "광고유입여부"
})
df_baemin["platform"] = "배민"
```

```python
df_coupang = pd.read_csv(coupang_file[0])
df_coupang = df_coupang.rename(columns={
    "order_time": "주문일시",
    "total_price": "전체금액",
    "advertisement_yn": "광고유입여부",
    "ad_pct": "광고비율"
})
df_coupang["platform"] = "쿠팡"
```

```python
df_total = pd.concat(
    [df_baemin, df_coupang],
    ignore_index=True
)
```

```python
df_pivot = df_total.pivot_table(
    index="주문일시",
    columns="platform",
    values="전체금액",
    aggfunc="sum"
)
df_pivot.isna().sum()
```

```python
df_total.groupby("광고유입여부")["전체금액"].sum()
```

```python
food_coat = bank[
    (bank["거래처"] == "식자재마트") &
    (bank["출금액"] > 0)
]
food_coat
```

```python
df_total["주문일시"] = pd.to_datetime(df_total["주문일시"])
df_total["주문일자"] = df_total["주문일시"].dt.date
```

```python
sales_table = df_total.pivot_table(
    index="주문일자",
    columns="platform",
    values="전체금액",
    aggfunc="sum"
    )
sales_table
```

```python
sales_table = sales_table.fillna(0)
sales_table["합계"] = sales_table["배민"] + sales_table["쿠팡"]
sales_table.head()
```

```python
final_table = sales_table.join(food_coat)
final_table
```

```python
final_table = sales_table.join(food_coat)
final_table = final_table.rename(
    columns={"출금액": "식자재비"}
)
```

```python
final_table["식자재비"] = final_table["식자재비"].fillna(0)
final_table["매출-식자재비"] = (
    final_table["합계"] - final_table["식자재비"]
)
final_table
```
