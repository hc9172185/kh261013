## 6. 실습 폴더 구성 — VS Code 탐색기

## 7. 데이터셋 다운로드 — seaborn 예제 데이터
seaborn에서 tips와 titanic 데이터를 내려받아 크기(행 수, 열 수)를 확인합니다.

```python
tips = sns.load_dataset("tips")              # 식당 팁 데이터 다운로드
titanic = sns.load_dataset("titanic")        # 타이타닉 승객 데이터 다운로드

print("tips   :", tips.shape)                # (행 수, 열 수)
print("titanic:", titanic.shape)
```

## 8. 데이터셋 저장 — CSV·Excel·JSON
tips는 CSV·Excel·JSON 세 형식으로, titanic은 CSV로 data/raw 폴더에 저장합니다.

```python
tips.to_csv("data/raw/tips.csv", index=False)           # CSV 저장
tips.to_excel("data/raw/tips.xlsx", index=False)        # Excel 저장
tips.to_json("data/raw/tips.json",                      # JSON 저장
             orient="records",                          # 행 하나 = 객체 하나
             force_ascii=False,                         # 한글 그대로 저장
             indent=2)                                  # 들여쓰기 2칸
titanic.to_csv("data/raw/titanic.csv", index=False)     # CSV 저장
print("저장 완료")
```
