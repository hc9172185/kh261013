## 3. Series — 이름표가 붙은 1차원 데이터
팁 값 3개로 Series를 만들고 값·인덱스를 꺼내 보고, 이름표로 값 하나를 꺼냅니다.

```python
kh = pd.Series(
    [1, 2, 3.14],
    index=["a","b","C"],
    name="kh"
)
print(kh)
```

```python
print(kh.values)
print(kh.index)
print(kh["a"])
```

## 4. DataFrame — Series를 옆으로 붙인 2차원 표
딕셔너리로 작은 DataFrame을 만들고 shape·columns·index와 열 하나의 타입을 확인합니다.

```python
x1 = pd.Series([10,20,30], name="첫번째시리즈")
x2 = pd.Series([40,50,60], name="두번째시리즈")
x3 = pd.Series([70,80,90], name="세번째시리즈")
df = pd.DataFrame(
    [x1,x2,x3]
)
df
```

```python
# 딕셔너리를 넣으면 초반에 넣은 name이 없앤다.
x1 = pd.Series([10,20,30], name="첫번째시리즈")
x2 = pd.Series([40,50,60], name="두번째시리즈")
x3 = pd.Series([70,80,90], name="세번째시리즈")
df = pd.DataFrame(
    {
        "one" : x1,
        "two" : [20, 21, 22],
        "three" : [100, 200, 300],
    }
)
# 둘다 가져오고 싶으면 대괄호를 쓴다.
# 대괄호를 혼자 쓰여도 되고 개수가 아니라 형태가 중요하다.
x = df[["one"]]
type(x)
```

## 5. CSV 불러오기 — read_csv
tips.csv를 불러와 크기와 앞 3행을 확인합니다.

```python
csv_df = pd.read_csv(
    "data/raw/tips.csv",                     # 파일 경로
    sep=",",                                 # 구분자 (기본값 ",")
    encoding="utf-8",                        # 문자 인코딩
)
print(csv_df.shape)
print(csv_df.head(10))                        # 앞 3행 미리보기
```

## 6. 필요한 열만 불러오기 - esecols
tips.csv에서 total_bill, tip, day 세 열만 골라 불러옵니다.

```python
csv_df = pd.read_csv(
    "data/raw/tips.csv",                     # 파일 경로
    sep=",",                                 # 구분자 (기본값 ",")
    encoding="utf-8",                        # 문자 인코딩  
    usecols=["total_bill", "tip", "day"]     
)
print(csv_df.shape)
print(csv_df.head(10))                       # 앞 3행 미리보기
```

## 7. Excel 불러오기 - read_excel
tips.xlsx의 첫번째 사진을 불러와 크기와 앞 3행을 확인합니다.

```python
excel_df = pd.read_excel(
    "data/raw/tips.xlsx",                     # 파일 경로
    sheet_name=0
)
print(excel_df.shape)
print(excel_df.head(10))                       # 앞 3행 미리보기
```

## 8. JSON 불러오기 - read_json
tips.json을 records 구조로 불러와 크기와 앞 3행을 확인합니다.

```python
json_df = pd.read_json(
    "data/raw/tips.json",                     # 파일 경로
    orient="records"
)
print(json_df.shape)
print(json_df.head(10))                       # 앞 3행 미리보기
```
